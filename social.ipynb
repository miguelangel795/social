RFC

```python
# Importar librerías necesarias
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
import matplotlib.pyplot as plt

# Cargar el conjunto de datos Digits
x = df.drop(['Affects_Academic_Performance'], axis=1)
y = df['Affects_Academic_Performance']

x = pd.get_dummies(x, columns=['Gender', 'Academic_Level', 'Country', 'Most_Used_Platform', 'Overall_Impact'], drop_first=True)



# Dividir el conjunto en entrenamiento (80%) y prueba (20%)
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

# Crear el modelo de Random Forest con 7 árboles
model = RandomForestClassifier(n_estimators=7, random_state=42)

# Entrenar el modelo0
model.fit(X_train, y_train)

# Hacer predicciones
y_pred = model.predict(X_test)

# Evaluar el modelo con accuracy
accuracy = accuracy_score(y_test, y_pred)
print(f'Precisión del Random Forest: {accuracy * 100:.2f}%')
```
